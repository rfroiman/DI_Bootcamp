## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.  
2) Identify key characteristics (what is open; what you can/cannot do).  
3) Compare side by side in a table (what's open vs. impact).  
4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```python

# Exercise 1: Definitions and key characteristics

open_source_levels = {
    "Fully Open": {
        "definition": (
            "A model whose architecture, source code, and pretrained "
            "weights are publicly available for inspection and modification."
        ),
        "what_is_open": [
            "Model architecture and implementation code",
            "Pretrained model weights"
        ],
        "what_you_can_do": [
            "Inspect and modify the model implementation",
            "Fine-tune or retrain the model, subject to its license"
        ],
        "what_you_cannot_do": [
            "Assume unrestricted commercial use without checking the license",
            "Assume training datasets are available unless explicitly released"
        ],
    },
    "Weights Released": {
        "definition": (
            "A model with publicly available pretrained weights, "
            "but without necessarily releasing its complete training code."
        ),
        "what_is_open": [
            "Pretrained model weights",
            "Model configuration and possibly inference code"
        ],
        "what_you_can_do": [
            "Run inference using the released weights",
            "Fine-tune the model when permitted by its license"
        ],
        "what_you_cannot_do": [
            "Fully inspect undisclosed training procedures",
            "Assume the original training process can be reproduced"
        ],
    },
    "Architecture Only": {
        "definition": (
            "A model whose architecture is publicly described, "
            "but whose pretrained weights are not released."
        ),
        "what_is_open": [
            "Model architecture or structural specifications",
            "Technical descriptions of model components"
        ],
        "what_you_can_do": [
            "Study the architecture and design",
            "Implement and train a similar model from scratch"
        ],
        "what_you_cannot_do": [
            "Use unavailable pretrained weights",
            "Fine-tune the original pretrained model"
        ],
    },
}

for level, details in open_source_levels.items():
    print(f"\n{level}")
    print(f"Definition: {details['definition']}")
    for category in [
        "what_is_open",
        "what_you_can_do",
        "what_you_cannot_do"
    ]:
        print(f"{category}:")
        for item in details[category]:
            print(f"  - {item}")
```

```python

# Exercise 1: Compare open-source LLM levels

comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Architecture, implementation code, and pretrained weights | Supports code inspection, modification, fine-tuning, and retraining, subject to license terms. |
| Weights Released | Pretrained weights and usually model configuration; full training code may be unavailable | Supports inference and potentially fine-tuning, but exact reproduction of original training may not be possible. |
| Architecture Only | Architecture specifications or technical descriptions, without pretrained weights | Allows studying or reimplementing the architecture, but requires training from scratch rather than fine-tuning the original model. |"""

print(comparison_table)
```

```python

# Exercise 1: Comparative paragraph

comparative_paragraph = (
    "Fully Open models provide access to their architecture, "
    "implementation code, and pretrained weights, allowing developers "
    "to inspect, modify, fine-tune, or retrain them. "
    "Weights Released models provide pretrained parameters that can "
    "support inference and fine-tuning, although the complete training "
    "process may not be available for inspection or reproduction. "
    "Architecture Only models disclose their structural design but "
    "do not provide pretrained weights, requiring developers to train "
    "their own implementations from scratch. "
    "Overall, the appropriate openness level depends on the need "
    "for transparency, customization, and available computing resources."
)

print(comparative_paragraph)
```
