# Daily Challenge: Precision Prompting for Control & Output Quality

*Week 9 - Day 5* | Developers Institute**

## Objective

Design, test, evaluate, and refine prompts to control the
tone, structure, length, and factual accuracy of AI-generated
employee communications.

## Scenario

Create employee-facing microlearning content based on an
internal remote access security policy.

## Workflow

1. Prompt Design: Tone, Format, and Length Control
2. Output Generation and Evaluation
3. Hallucination Detection and Mitigation
4. Paraphrasing for Junior Interns
5. Direct Quote Extraction and Risk Analysis

```python
# Original remote access security policy

policy_text = """
Employees must ensure that all remote access to internal systems
is established via the approved secure VPN. Under no circumstances
should unsecured connections or personal devices lacking endpoint
protection be used to access proprietary data or sensitive communications.
"""

print(policy_text)
```

```python
# Step 1: Design a prompt with tone, format, and length constraints

prompt_v1 = f"""
Act as an internal communications specialist at a global
technology company.

Rewrite the following remote access security policy into
a short microlearning snippet for an employee newsletter.

Requirements:
1. Use a friendly, clear, and professional tone.
2. Present the information using bullet points.
3. Paraphrase the original text without direct quotations.
4. Keep the entire response under 75 words.
5. Use simple language suitable for non-technical employees.
6. Preserve all critical security requirements.
7. Do not introduce new security policies or technologies.

Original Policy:
{policy_text}

Output only the rewritten microlearning snippet.
"""

print(prompt_v1)
```

```python
# Step 2: Store and validate the AI-generated response

generated_response_v1 = """
- When working remotely, always connect to company systems through the approved secure VPN.
- Avoid unsecured connections when accessing company information.
- Only use personal devices with proper security protection when handling sensitive messages or confidential company data.
"""

word_count = len(generated_response_v1.split())
bullet_count = sum(
    line.strip().startswith("- ")
    for line in generated_response_v1.splitlines()
)

print("Generated Response:")
print(generated_response_v1)

print(f"\nWord Count: {word_count}")
print(f"Bullet Points: {bullet_count}")
print(f"Under 75 Words: {word_count < 75}")
```

```python
# Step 2: Evaluate the generated response

evaluation_v1 = {
    "Relevance": {
        "passed": True,
        "notes": "The response focuses on remote access security and preserves the main policy requirements."
    },
    "Clarity": {
        "passed": True,
        "notes": "The language is concise and understandable for non-technical employees."
    },
    "Structure": {
        "passed": True,
        "notes": "The response uses three clear bullet points."
    },
    "Tone": {
        "passed": True,
        "notes": "The tone is friendly, professional, and appropriate for an internal newsletter."
    },
    "Length": {
        "passed": word_count < 75,
        "notes": f"The response contains {word_count} words, below the 75-word limit."
    },
    "Factual Accuracy": {
        "passed": True,
        "notes": (
            "The response preserves the VPN requirement, prohibition "
            "of unsecured connections, and endpoint protection requirement "
            "for personal devices. No unsupported security policies were added."
        )
    }
}

for criterion, result in evaluation_v1.items():
    status = "PASS" if result["passed"] else "FAIL"
    print(f"{criterion}: {status}")
    print(f"  {result['notes']}\n")
```

```python
# Step 3: Refine the prompt to prevent hallucinations
# and preserve critical security requirements.

prompt_v2 = f"""
Act as an internal communications specialist responsible
for creating accurate employee security awareness content.

Rewrite the following remote access security policy
for an internal employee newsletter.

STRICT CONTENT RULES:
1. Use only information explicitly stated in the source.
2. Do not add new security policies, recommendations,
   technologies, procedures, or exceptions.
3. Preserve all mandatory security requirements:
   - Remote access must use the approved secure VPN.
   - Unsecured connections must not be used.
   - Personal devices without endpoint protection
     must not access proprietary data or sensitive communications.
4. Do not weaken mandatory restrictions or change their meaning.
5. Paraphrase the policy without direct quotations.

FORMAT AND STYLE:
- Use a friendly, clear, and professional tone.
- Write exactly 3 bullet points.
- Keep the entire response under 75 words.
- Use simple language suitable for non-technical employees.

VERIFICATION:
Before responding, check every statement against
the original policy text.

If a statement introduces unsupported information,
remove it.

Ensure that all mandatory requirements remain intact.

Original Policy:
{policy_text}

Output only the final 3 bullet points.
"""

print(prompt_v2)
```

```python
# Step 3: Generate and validate the refined response

generated_response_v2 = """
- Always use the company-approved secure VPN when accessing internal systems remotely.
- Never use unsecured connections to access confidential company data or sensitive communications.
- Do not use personal devices without endpoint protection to access proprietary data or sensitive communications.
"""

word_count_v2 = len(generated_response_v2.split())

bullet_count_v2 = sum(
    line.strip().startswith("- ")
    for line in generated_response_v2.splitlines()
)

required_checks = {
    "Approved secure VPN": "secure VPN" in generated_response_v2,
    "Unsecured connections prohibited": "Never use unsecured connections" in generated_response_v2,
    "Endpoint protection requirement": "without endpoint protection" in generated_response_v2,
    "Proprietary data protected": "proprietary data" in generated_response_v2,
    "Sensitive communications protected": "sensitive communications" in generated_response_v2,
    "Exactly 3 bullet points": bullet_count_v2 == 3,
    "Under 75 words": word_count_v2 < 75
}

print("Refined Response:")
print(generated_response_v2)

print(f"\nWord Count: {word_count_v2}")
print(f"Bullet Points: {bullet_count_v2}")

print("\nValidation Results:")
for requirement, passed in required_checks.items():
    print(f"{requirement}: {'PASS' if passed else 'FAIL'}")

print("\nOverall Validation:",
      "PASS" if all(required_checks.values()) else "FAIL")
```

```python
# Step 4: Create a simplified prompt for junior interns

intern_prompt = f"""
Act as a friendly employee training specialist.

Rewrite the following remote access security policy
for junior interns who are new to the company.

AUDIENCE:
- Junior interns with little technical knowledge.
- Employees who may be unfamiliar with security terminology.

INSTRUCTIONS:
1. Explain the policy using plain, everyday English.
2. Use short, simple sentences.
3. Write no more than 4 bullet points.
4. Use a supportive and informative tone.
5. Avoid corporate, legal, and unnecessary technical jargon.
6. Paraphrase the original policy without direct quotations.
7. Preserve all mandatory security requirements.
8. Do not add new rules, exceptions, or technologies.

IMPORTANT SECURITY REQUIREMENTS:
- Remote access to internal systems must use
  the approved secure VPN.
- Unsecured connections are prohibited when
  accessing proprietary data or sensitive communications.
- Personal devices without endpoint protection
  must not be used for such access.

CLARITY RULES:
- Explain technical terms in simple language when necessary.
- Do not weaken or change the meaning of the policy.
- Keep the entire response under 75 words.

Original Policy:
{policy_text}

Output only the simplified bullet points.
"""

print(intern_prompt)
```

```python
# Step 4: Generate and validate the simplified policy

intern_response = """
- Working from home? Always use the company's approved secure VPN to access internal systems.
- Never use an unsafe connection to access private company information or sensitive messages.
- Do not use personal devices without security software that protects them from threats.
- These rules help keep company information and communications safe.
"""

word_count_intern = len(intern_response.split())

bullet_count_intern = sum(
    line.strip().startswith("- ")
    for line in intern_response.splitlines()
)

intern_validation = {
    "Maximum 4 bullet points": bullet_count_intern <= 4,
    "Under 75 words": word_count_intern < 75,
    "VPN requirement included": "VPN" in intern_response,
    "Unsecured connections prohibited": "unsafe connection" in intern_response,
    "Device protection included": "security software" in intern_response
}

print("Simplified Policy for Junior Interns:")
print(intern_response)

print(f"\nWord Count: {word_count_intern}")
print(f"Bullet Points: {bullet_count_intern}")

print("\nValidation Results:")
for requirement, passed in intern_validation.items():
    print(f"{requirement}: {'PASS' if passed else 'FAIL'}")

print("\nOverall Validation:",
      "PASS" if all(intern_validation.values()) else "FAIL")
```

```python
# Step 5: Create a prompt for extracting a direct policy quote

quote_prompt = f"""
Act as an internal compliance communications specialist.

Extract exactly one direct quote from the original
remote access security policy.

INSTRUCTIONS:
1. Select the sentence or passage that best communicates
   the core remote access security requirement.
2. Copy the selected text exactly as written.
3. Do not paraphrase, summarize, or modify the wording.
4. Do not invent or combine unrelated passages.
5. Preserve the original meaning and terminology.
6. Return only one direct quote enclosed in quotation marks.

Original Policy:
{policy_text}

Output only the exact quotation.
"""

print(quote_prompt)
```

```python
# Step 5: Extract and validate an exact quotation

extracted_quote = (
    "Employees must ensure that all remote access to internal systems "
    "is established via the approved secure VPN."
)

normalized_policy = " ".join(policy_text.split())
normalized_quote = " ".join(extracted_quote.split())

quote_validation = {
    "Exact quotation": normalized_quote in normalized_policy,
    "Core security requirement": "approved secure VPN" in extracted_quote,
    "Single quotation": True,
    "No fabricated content": normalized_quote in normalized_policy
}

print("Extracted Quote:")
print(f'"{extracted_quote}"')

print("\nValidation Results:")
for criterion, passed in quote_validation.items():
    print(f"{criterion}: {'PASS' if passed else 'FAIL'}")

print("\nOverall Validation:",
      "PASS" if all(quote_validation.values()) else "FAIL")
```

```python
# Step 5: Explain appropriate quotation use and potential risks

quote_analysis = {
    "When quoting is appropriate": (
        "Direct quotations are appropriate in official policy "
        "announcements, compliance training, security reminders, "
        "and audit documentation where preserving the exact "
        "wording of mandatory requirements is important."
    ),
    "When quoting poses risks": (
        "Quoting can create confusion when the original language "
        "is too technical for the audience, when a passage is "
        "taken out of context, or when an outdated policy "
        "version is presented as current. Incomplete quotations "
        "may also omit critical restrictions or exceptions."
    )
}

for question, answer in quote_analysis.items():
    print(f"{question}:\n{answer}\n")
```
