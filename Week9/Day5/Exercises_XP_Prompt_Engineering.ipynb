## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
# Identify three critical issues with the vague prompt
critical_issues = [
    "The prompt does not define the target audience, such as professionals, students, or remote workers who want to improve productivity.",
    "The prompt lacks brand context and voice, failing to mention FlowNest, its purpose, or how the LinkedIn post should promote the app.",
    "The prompt provides no instructions about content structure, tone, length, actionable tips, or engagement strategies appropriate for LinkedIn."
]

critical_issues
```

## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# Create a structured educational prompt using role prompting,
# clear instructions, audience specification, and output constraints.

quiz_prompt = """
Act as an experienced 7th-grade science teacher and educational
content designer.

Using only the provided article about volcanic eruptions, create
engaging educational content for students aged 11 to 13.

Target audience:
7th-grade students with basic scientific knowledge.

Task 1 - Article Summary:
- Summarize the article in exactly 2 bullet points.
- Highlight the most important scientific concepts.
- Use simple, age-appropriate vocabulary.

Task 2 - Multiple-Choice Quiz:
- Generate exactly 3 multiple-choice questions.
- Each question must have exactly 3 answer options (A, B, C).
- Include 1 correct answer and 2 plausible distractors.
- Base every question and answer on the provided article.
- Clearly identify the correct answer in a separate answer key.

Tone and Language:
- Use a friendly, encouraging, and educational tone.
- Avoid unnecessary scientific jargon.
- Explain essential scientific terms in simple language.

Output Format - Google Slides:
Slide 1: Title - Understanding Volcanic Eruptions
Slide 2: Article Summary - Exactly 2 bullet points
Slide 3: Multiple-Choice Question 1
Slide 4: Multiple-Choice Question 2
Slide 5: Multiple-Choice Question 3
Slide 6: Answer Key - Correct answers with brief explanations

Additional Constraints:
- Keep each slide concise and easy to read.
- Use clear headings and short sentences.
- Do not introduce facts that are not supported by the article.
- If the article lacks sufficient information, state this
  instead of inventing additional facts.

Article:
[PASTE ARTICLE HERE]
"""

print(quiz_prompt)
```

```python
# Explain three improvements over the vague prompt.

improvements = [
    "Clear role and audience: The improved prompt assigns the model the role of a 7th-grade science teacher and specifies students aged 11 to 13, ensuring appropriate vocabulary, explanations, and educational difficulty.",

    "Specific instructions and content constraints: Unlike the vague prompt, it requires exactly 2 summary bullet points and 3 multiple-choice questions with 1 correct answer and 2 distractors each, reducing ambiguity and incomplete responses.",

    "Structured output and factual accuracy: The prompt defines a six-slide Google Slides format, includes an answer key, and requires all information to come from the provided article, improving presentation consistency and reducing hallucination risk."
]

improvements
```

## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
# Choose a prompt style and create a customer support template.

chosen_style = 'Structured'

prompt_template = """
Act as a professional customer support assistant for an
e-commerce company.

Your role is to help customers resolve issues related to
orders, deliveries, returns, refunds, and product information.

Follow this structured response format:

1. Greeting and Acknowledgment:
   - Greet the customer politely.
   - Acknowledge the customer's question or concern.
   - Show empathy when the customer reports a problem.

2. Issue Identification:
   - Identify the customer's main request.
   - Ask for missing information only when necessary.
   - Never request passwords or payment card details.

3. Solution or Next Steps:
   - Provide clear, actionable instructions.
   - Use numbered steps when explaining procedures.
   - Offer alternatives when appropriate.
   - Do not invent order statuses, policies, or refund dates.

4. Confirmation and Closing:
   - Summarize the recommended solution.
   - Ask whether the customer needs further assistance.
   - End with a friendly and professional closing.

Tone and Style:
- Be polite, empathetic, and solution-oriented.
- Use simple, concise, and professional language.
- Avoid unnecessary technical jargon.
- Keep responses under 150 words whenever possible.

Important Constraints:
- Use only verified order information and company policies.
- If information is unavailable, explain the limitation.
- Escalate unresolved or sensitive issues to a human agent.
- Never promise refunds or delivery dates without confirmation.

Customer Message:
[INSERT CUSTOMER MESSAGE HERE]
"""

key_style_features = [
    "Consistent response structure: The prompt defines four ordered sections covering acknowledgment, issue identification, resolution, and closing.",

    "Explicit formatting and behavioral rules: The chatbot follows clear instructions for response length, tone, step-by-step solutions, and escalation procedures."
]

justification = """
The Structured style is the most suitable choice for an
e-commerce customer support chatbot because it ensures
consistent, organized, and actionable responses.

Unlike the Exploratory style, which focuses on generating
ideas, the Structured style prioritizes resolving customer
issues through predefined steps.

Compared with the Conversational style, it provides stronger
consistency and makes important information easier to find.

Although the Functional style is effective for executing
specific tasks, the Structured style is more flexible for
handling different customer support scenarios while
maintaining a predictable response format.

This approach improves clarity, service quality, and
customer satisfaction.
"""

print("Chosen Style:", chosen_style)
print("\nPrompt Template:\n", prompt_template)
print("\nKey Style Features:")
for i, feature in enumerate(key_style_features, 1):
    print(f"{i}. {feature}")
print("\nJustification:\n", justification)
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# Create a source-grounded prompt to reduce hallucinations.

no_hallucination_prompt = """
Act as a scientific research assistant specializing in
climate change and marine biodiversity.

Summarize the provided peer-reviewed scientific article
using only the information explicitly contained in the text.

Instructions:
1. Identify the article's main findings.
2. Summarize the key impacts of climate change on
   marine biodiversity discussed in the article.
3. Include numerical data only when explicitly
   provided in the source.
4. Do not introduce external facts, assumptions,
   predictions, or unsupported conclusions.
5. Preserve the original meaning and scientific context.

Output Format:
- Main Findings: 3 concise bullet points.
- Supporting Evidence: Relevant findings from the article.
- Conclusion: A brief, evidence-based summary.

Important:
Do not invent statistics, extinction projections,
scientific findings, or references.

If information is missing, acknowledge the limitation
instead of generating unsupported claims.

Scientific Article:
[PASTE ARTICLE HERE]
"""

print(no_hallucination_prompt)
```

```python
# Strengthen the prompt with explicit verification requirements.

verification_prompt = """
Act as a scientific research analyst responsible for
producing accurate, evidence-based article summaries.

Analyze the provided peer-reviewed article about
climate change and marine biodiversity.

STRICT VERIFICATION RULES:

1. Use only claims explicitly stated in the article.
2. Verify every factual statement against the source text.
3. Do not introduce external knowledge or assumptions.
4. Do not invent statistics, percentages, dates,
   extinction estimates, or scientific conclusions.
5. Preserve uncertainty and limitations described
   by the original researchers.
6. If a claim cannot be verified, exclude it or state:
   "This information is not available in the article."

Specific Verification Example:
Do not include the statement
"Over 50% of marine species are projected to go
extinct by 2050"
unless the article explicitly supports this claim.

Required Output:

Section 1 - Verified Summary:
- Present exactly 3 evidence-based findings.
- Use clear, objective scientific language.

Section 2 - Supporting Evidence:
- Provide a short supporting quotation or passage
  from the article for each finding.
- Do not fabricate quotations or references.

Section 3 - Limitations:
- Identify missing information or uncertainties.
- Clearly distinguish findings from interpretations.

Final Verification:
Before responding, check that every factual claim
is supported by the provided article.
Remove or correct any unsupported statement.

Scientific Article:
[PASTE ARTICLE HERE]
"""

print(verification_prompt)
```

```python
# Identify hallucination mitigation strategies and high-risk domains.

mitigation_strategies = [
    "Source-grounded generation: Restrict the model to information explicitly contained in the provided article, preventing unsupported external claims, invented statistics, and speculative conclusions.",

    "Evidence-based verification: Require the model to verify factual statements against the source, provide supporting evidence, and explicitly acknowledge missing or uncertain information."
]

risky_domains = {
    'domain_1': {
        'name': 'Healthcare',
        'why': (
            "Hallucinated medical information can lead to incorrect "
            "diagnoses, inappropriate treatments, medication errors, "
            "and serious risks to patient safety."
        )
    },
    'domain_2': {
        'name': 'Legal Services',
        'why': (
            "Fabricated legal precedents, incorrect interpretations "
            "of laws, or invented regulations can mislead legal "
            "professionals and clients, resulting in financial "
            "losses, invalid legal arguments, or judicial penalties."
        )
    }
}

print("Mitigation Strategies:")
for i, strategy in enumerate(mitigation_strategies, 1):
    print(f"{i}. {strategy}")

print("\nHigh-Risk Professional Domains:")
for domain in risky_domains.values():
    print(f"\nDomain: {domain['name']}")
    print(f"Why: {domain['why']}")
```
